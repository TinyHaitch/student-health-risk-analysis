```python
## General Purpose
import warnings
from itertools import combinations
from copy import deepcopy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

# ── Step 1: Load both files correctly ──────────────
train = pd.read_csv("data/train.csv")
test  = pd.read_csv("data/test.csv")

print('Train shape:', train.shape)
print('Test shape: ', test.shape)

# ── Step 2: Drop id — not a feature ────────────────
train = train.drop('id', axis=1).copy()
test  = test.drop('id', axis=1).copy()

# ── Step 3: Separate column types ──────────────────
num_cols = train.select_dtypes(
    include=['float64', 'int64']
).columns.tolist()

cat_cols = train.select_dtypes(
    include='str'
).columns.tolist()

# Remove target from both lists
cat_cols = [col for col in cat_cols if col != 'health_condition']

print('\nNumeric columns:', num_cols)
print('Categorical columns:', cat_cols)
print('Target: health_condition')

# ── Step 4: Check missing values ───────────────────
print('\n=== MISSING VALUES (%) ===')
print((train.isnull().sum() / len(train) * 100).round(2))

# ── Step 5: Check target distribution ──────────────
print('\n=== TARGET DISTRIBUTION ===')
print(train['health_condition'].value_counts())
print(train['health_condition'].value_counts(
    normalize=True).round(3) * 100) 
```

```python
print('=== SKEWNESS ===')
print(train[num_cols].skew().round(3).sort_values(ascending=False))

print('\n=== BASIC STATS ===')
print(train[num_cols].describe().round(2))
```

```python

print('\n=== CATEGORICAL DISTRIBUTIONS ===')
for col in cat_cols:
    print(f'\n{col}:')
    print(train[col].value_counts())
    print(f'Missing: {train[col].isnull().sum()} ({train[col].isnull().sum()/len(train)*100:.1f}%)')
```
